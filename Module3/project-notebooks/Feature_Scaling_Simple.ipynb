```python
import numpy as np

# Two points
A = np.array([1000, 8])
B = np.array([2000, 7])

# Distance without scaling
distance = np.sqrt((A[0] - B[0])**2 + (A[1] - B[1])**2)

print("Distance without scaling:", distance)
```

```python
# Scale the features
from sklearn.preprocessing import MinMaxScaler

data = np.array([
    [1000, 8],
    [2000, 7]
])

scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(data)

A_scaled = scaled_data[0]
B_scaled = scaled_data[1]

print("Scaled A:", A_scaled)
print("Scaled B:", B_scaled)
```

```python
# Distance after scaling
scaled_distance = np.sqrt(
    (A_scaled[0] - B_scaled[0])**2 +
    (A_scaled[1] - B_scaled[1])**2
)

print("Distance after scaling:", scaled_distance)
```

```python
# Compare distances
print("Without scaling:", round(distance, 4))
print("With scaling   :", round(scaled_distance, 4))
```
