# MNIST Handwritten Digit Classifier Using ANN

```python
import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np
```

### 1. Load the MNIST dataset

```python
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()

print('Training data:', X_train.shape)
print('Testing data:', X_test.shape)
```

### 2. View sample images

```python
plt.figure(figsize=(8, 4))
for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(X_train[i], cmap='gray')
    plt.title(f'Label: {y_train[i]}')
    plt.axis('off')
plt.tight_layout()
plt.show()
```

### 3. Preprocess the images

```python
X_train = X_train.astype('float32') / 255.0
X_test = X_test.astype('float32') / 255.0
```

### 4. Build the ANN model

```python
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(28, 28)),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.summary()
```

### 5. Train the model

```python
history = model.fit(
    X_train, y_train,
    epochs=5,
    batch_size=32,
    validation_split=0.1
)
```

### 6. Evaluate the model

```python
test_loss, test_accuracy = model.evaluate(X_test, y_test, verbose=0)
print(f'Test accuracy: {test_accuracy:.4f}')
```

### 7. Plot training accuracy

```python
plt.plot(history.history['accuracy'], label='Training accuracy')
plt.plot(history.history['val_accuracy'], label='Validation accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.show()
```

### 8. Predict a handwritten digit

```python
index = 0
prediction = model.predict(X_test[index:index + 1], verbose=0)
predicted_digit = np.argmax(prediction)

plt.imshow(X_test[index], cmap='gray')
plt.title(f'Predicted: {predicted_digit} | Actual: {y_test[index]}')
plt.axis('off')
plt.show()

print('Class probabilities:', np.round(prediction[0], 3))
```

### 9. Save the trained model

```python
model.save('mnist_ann_classifier.keras')
```
